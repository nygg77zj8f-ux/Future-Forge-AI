```python
def show_title():
    print("=" * 50 )
    print(" Future Forge AI")
    print("AI-Powered Portfolio & Career Mentor")
    print("=" * 50 )

def get_user_information():

    name=input("Enter your name:")

    print()

    print("Welcome,",name)

    gpa=float(input("Enter your GPA (0.00-4.00):"))

    english_level=input("Enter your English Level (A1,A2,B1,B2,C1):")

    github_projects=int(input("Enter your GitHub Project count:"))

    certificate_count=int(input("Enter your Certificate count:"))

    return name,gpa,english_level,github_projects,certificate_count


def calculate_score(gpa,english_level,github_projects,certificate_count):

    score = 0

    if gpa >= 3.50:
       score += 24
    elif gpa >=3.00:
       score += 12
    else:
       score += 6

    if english_level == "C1":
       score += 18
    elif english_level == "B2":
       score += 9
    elif english_level == "B1":
       score += 5
    else:
       score += 2

    if github_projects >= 6:
       score += 25
    elif github_projects >= 3:
       score += 18
    else:
       score += 7

    if certificate_count >= 5:
       score += 12
    elif certificate_count >= 3:
       score += 6
    elif certificate_count >= 1:
       score += 4
    else:
       score += 1

    return score

def show_report(name,score):

    print()
    print("=" * 50 )
    print("Portfolio Analysis Report")
    print("=" * 50 )

    print("Student:",name)
    print("Portfolio Score:",score,"/100")

    if score >= 80:
          print("Excellent Portfolio!:")
    elif score >= 60:
          print("Good Portfolio!:")
    else:
          print("You should improve your portfolio.")

def show_suggestions (gpa,english_level,github_projects,certificate_count):

        print()
        print("Suggestions")
        print("=" * 50 )
        if gpa < 3.50:
           print("-Improve your GPA:")
        if english_level == "A1" or english_level == "A2" or english_level == "B1":
           print("-Improve your English level:")
        if github_projects < 3:
           print("-Build more GitHub projects:")
        if certificate_count < 2:
           print("-Earn more certificates:")

def main():
        show_title()

        name,gpa,english_level,github_projects,certificate_count = get_user_information()

        score = calculate_score (gpa,english_level,github_projects,certificate_count)

        show_report(name,score)

        show_suggestions(gpa,english_level,github_projects,certificate_count)

if __name__ == "__main__":
        main()








```
